```python
twitter = pd.read_csv('./twcs.csv')
```

```python
twitter
```

```python
X = twitter.text
```

```python
def is_company(srs):
  try:
    author_id = int(srs)
    return 0
  except ValueError as e:
    if isinstance(srs, str) or isinstance(srs, bytes):
      return 1
    else:
      raise ValueError(f'{srs} is something different')

twitter['is_company'] = twitter.author_id.apply(is_company)
```

```python
y = twitter.is_company
```

```python
import re
nltk.download('stopwords')
my_stopwords = nltk.corpus.stopwords.words('english')
word_rooter = nltk.stem.snowball.PorterStemmer(ignore_stopwords=False).stem
my_punctuation = '!"$%&\'()*+,-./:;<=>?[\\]^_`{|}~•@'


def preprocess_text(text, should_join=True):
    text = ' '.join(word.lower() for word in textblob_tokenizer(text))
    text = re.sub(r'http\S+', '', text) # remove http links
    text = re.sub(r'bit.ly/\S+', '', text) # rempve bitly links
    text = text.strip('[link]') # remove [links]
    text = re.sub(r"(\@\d+)", r" ", text) # remove twitter handle
    text = re.sub(r"(https\:\/\/t.co\/\w+)", r" ", text) # remove end twtter url
    text = re.sub('['+my_punctuation + ']+', ' ', text) # remove punctuation
    text = re.sub('\s+', ' ', text) #remove double spacing
    text = re.sub(r"[^a-zA-Z.,&!?]+", r" ", text) # only normal characters
    text_token_list = [word for word in text.split(' ')
                            if word not in my_stopwords] # remove stopwords
    text_token_list = [word_rooter(word) if '#' not in word else word
                        for word in text_token_list] # apply word rooter
    text = ' '.join(text_token_list)
    if should_join:
      return ' '.join(gensim.utils.simple_preprocess(text))
    else:
      return gensim.utils.simple_preprocess(text)
```

```python
y.value_counts()
```

```python
sampled_x = X.sample(frac=sample_size)
```

```python
sampled_y = y[sampled_x.index]
```

```python
sampled_y.value_counts()
```

```python
import swifter
X_preprocessed = sampled_x.swifter.apply(preprocess_text)
```

```python
X_preprocessed.iloc[0]
```

```python
# Find a set named vocab that has all unique words
from collections import defaultdict
vocab = set()
counts = defaultdict(int)
min_counts_words = 3

def update_vocab(review):
  for word in preprocess_text(review, should_join=False):
    counts[word] += 1
    if counts[word] > min_counts_words:
      vocab.add(word)

X_preprocessed.swifter.apply(update_vocab)
vocab=sorted(vocab)   # this is not needed, this is my TOC
```

```python
maximum = 0
def get_maximum_review_length(review):
  global maximum
  candidate = len(preprocess_text(review, should_join=False))
  if candidate > maximum:
    maximum = candidate


X_preprocessed.swifter.apply(get_maximum_review_length)
```

```python
import tensorflow as tf
from tensorflow.keras.layers.experimental import preprocessing
ids_from_words = preprocessing.StringLookup(vocabulary=list(vocab), mask_token=None)
```

```python
words_from_ids = preprocessing.StringLookup(
    vocabulary=ids_from_words.get_vocabulary(), invert=True, mask_token=None)
```

```python
def pad_sequence_of_tokens(x, maxlen, unk_token='[UNK]'):
  if len(x)<maxlen:
    x.extend([unk_token]*(maxlen-len(x)))
  return x
```

```python
from keras_preprocessing.sequence import pad_sequences
def get_ids_tensor(srs):

  processed = srs.swifter.apply(lambda x: pad_sequence_of_tokens(preprocess_text(x, should_join=False), maxlen=maximum)).to_list()
  return tf.squeeze(tf.constant(pad_sequences(ids_from_words(processed), maxlen=maximum, padding='post'), dtype='int32'))

```

```python
all_ids = get_ids_tensor(srs=X_preprocessed.reset_index(drop=True))
all_ids
```

```python
all_ids.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(all_ids.numpy(), sampled_y, test_size=0.25 ,random_state=99)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logreg = LogisticRegression()
logreg.fit(X_train, y_train)

```

```python
from sklearn import metrics
y_pred = logreg.predict(X_test)
print((metrics.accuracy_score(y_test, y_pred)))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

Now with NN

```python
import tensorflow as tf
import sys
import keras.backend as K
from keras.models import Sequential
from keras.layers import Dense, Embedding, Lambda
from keras.preprocessing import sequence
from keras.preprocessing.text import Tokenizer
from textblob import TextBlob, Word
from keras_preprocessing.sequence import pad_sequences
from keras.initializers import Constant
```

```python
model = Sequential()
model.add(Embedding(input_dim=all_ids.shape[1], output_dim=100, input_length=maximum, trainable=True))
model.add(Dense(100, activation='relu'))
model.add(Dense(50, activation='relu'))
model.add(Lambda(lambda x: K.mean(x, axis=1), output_shape=(None, 50,)))
model.add(Dense(50, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
precision = tf.keras.metrics.Precision()
recall = tf.keras.metrics.Recall()

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy', precision, recall])
model.summary()
```

```python
callback = tf.keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=5, min_delta=0.01)
history = model.fit(x=X_train, y=y_train, epochs=100, batch_size=500, validation_data=(X_test, y_test), workers=5, callbacks=[callback])
```

```python
import matplotlib.pyplot as plt

# function for plotting loss
def plot_metrics(train_metric, val_metric=None, metric_name=None, title=None, ylim=5):
    plt.title(title)
    plt.ylim(0,ylim)
    plt.plot(train_metric,color='blue',label=metric_name)
    if val_metric is not None: plt.plot(val_metric,color='green',label='val_' + metric_name)
    plt.legend(loc="upper right")

# plot loss history
plot_metrics(history.history['loss'], history.history['val_loss'], "Loss", "Loss", ylim=2.0)
```

```python
model.evaluate(X_test, y_test)
```
